### Why Visualize?

Our brains are wired to process visual information far more effectively than tables of numbers. Good plots can instantly reveal:

- **Trends:** How does a value change over time?
- **Comparisons:** How do different groups compare against each other?
- **Distributions:** How is our data spread out? Are there outliers?
- **Relationships:** Do two variables seem to be related?

Data visualization is the graphical representation of information and data. By using visual elements like charts, graphs, and maps, visualization tools provide an accessible way to see and understand trends, outliers, and patterns in data.

## 📊 Quick Decision Guide

|------------|----------------------------------|-------------------------------------------|

```python

# introduces grid plt.suptitle
# plt.suptitle() = the heading at the very top of the page, like "Chapter 5: Data Visualization"
# plt.title() = the smaller heading above each individual exercise/question on that page

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Sample data - Monthly sales over a year
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales_2023 = [120, 135, 150, 165, 180, 200, 
              190, 210, 220, 230, 250, 300]
sales_2024 = [130, 145, 160, 170, 185, 210, 
              200, 220, 235, 245, 265, 320]

sales_2025 = [100, 1450, 180, 800, 100, 220, 
              200, 290, 235, 205, 400, 1320]

plt.figure(figsize=(12, 6))
plt.plot(months, sales_2023, marker='o', linewidth=2, label='2023 Sales', color='blue')
plt.plot(months, sales_2024, marker='s', linewidth=2, label='2024 Sales', color='red', linestyle=':')
plt.plot(months, sales_2025, marker='D', linewidth=2, label='2025 Sales', color='green', linestyle='--')

plt.suptitle('Monthly Sales Comparison (2023 vs 2024 vs 2025)', fontsize=14, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Sales (in thousands $)', fontsize=12)
plt.grid(True, alpha=0.9)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Multiple subplots for different metrics
_, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Simple line plot
axes[0, 0].plot(months, sales_2023)
axes[0, 0].set_title('Simple Line Plot')

# 2. Area plot (filled under line)
axes[0, 1].fill_between(months, sales_2023, alpha=0.3)
axes[0, 1].set_title('Area Plot')
# fill_between() is new. Instead of just drawing a line, it shades in the area underneath 
# the line, down to the x-axis — like coloring in the space between the line and the ground.

# 3. Step plot
axes[1, 0].step(months, sales_2023, where='mid')
axes[1, 0].set_title('Step Plot')

# step() is new too. Instead of connecting points with a smooth diagonal line (like plot() 
# does), it draws the connection as sharp horizontal-then-vertical "staircase" jumps.
# This is useful when a value stays constant for a while, then jumps suddenly — like a 
# price that stays fixed for a month then jumps to a new fixed price, rather than gradually
# drifting upward.
# where='mid' controls exactly where the step/jump happens relative to each data point.
# 'mid' means the vertical jump happens midway between two x-positions (rather than exactly 
# at the point, or right before it) — this is a small visual/technical detail about staircase 
# alignment; the main thing to remember for now is that step() gives you a blocky staircase 
# shape instead of a smooth line.

# 4. Stacked area plot
axes[1, 1].stackplot(months, sales_2023, sales_2024, sales_2025,
                      labels=['2023', '2024','2025'], alpha=0.7)
axes[1, 1].set_title('Stacked Area Plot')
axes[1, 1].legend(loc='upper right')
# stackplot() is new and a bit different from everything else. Instead of drawing 3 separate
# lines for sales_2023, sales_2024, sales_2025 independently, it stacks them on top of each
# other — like piling one dataset's area on top of the previous one's area, so the total 
# height at any point represents the sum of all three years combined.

plt.tight_layout()
plt.show()
```

```python
# A stacked area chart is useful when you want to see two things at once:

# The overall trend/total — the height of the entire stack shows you the combined total (2023+2024+2025) at each month.
# Each year's individual share — the thickness of each colored band shows you how much that particular year contributed to the total at that point.

# Why not just use 3 separate line plots instead?

# With 3 separate lines, you could compare each year individually pretty easily, but you'd have no easy way to see the combined total — you'd have to mentally add up 3 different line heights yourself, which is hard to do by eye.
# With a stacked area chart, the "total" is already visually built for you — no mental math needed.

# One-line takeaway: Use a stacked area chart when you care about both the grand total and each part's contribution to that total — not just comparing trends side-by-side.

```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Sample data
categories = ['A', 'B', 'C', 'D', 'E']
values1 = [25, 40, 30, 35, 20]
values2 = [30, 35, 10, 40, 30]

# 1. Horizontal bar plot
axes[0, 0].barh(categories, values1, color='lightblue')
axes[0, 0].set_title('Horizontal Bar Plot')


# 2. Stacked bar plot
axes[0, 1].bar(categories, values1, label='Group 1', color='lightblue')
axes[0, 1].bar(categories, values2, bottom=values1, 
               label='Group 2', color='lightcoral')
axes[0, 1].set_title('Stacked Bar Plot')
axes[0, 1].legend()


# 3. Grouped bar plot
x = np.arange(len(categories))
width = 0.35
axes[1, 0].bar(x - width/2, values1, width, label='Group 1')
axes[1, 0].bar(x + width/2, values2, width, label='Group 2')
axes[1, 0].set_xticks(x)
axes[1, 0].set_xticklabels(categories)
axes[1, 0].set_title('Grouped Bar Plot')
axes[1, 0].legend()


# 4. Percentage bar plot
total = [v1 + v2 for v1, v2 in zip(values1, values2)]
pct1 = [v1/t*100 for v1, t in zip(values1, total)]
pct2 = [v2/t*100 for v2, t in zip(values2, total)]
axes[1, 1].bar(categories, pct1, label='Group 1 %')
axes[1, 1].bar(categories, pct2, bottom=pct1, label='Group 2 %')
axes[1, 1].set_title('Percentage Bar Plot')
axes[1, 1].set_ylabel('Percentage (%)')
axes[1, 1].legend()

plt.tight_layout()
plt.show()
```
