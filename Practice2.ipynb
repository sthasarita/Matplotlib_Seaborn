```python

# demo of the use of plt.tight_layout() method


import matplotlib.pyplot as plt

product_names = [
    "Wireless Bluetooth Headphones",
    "Stainless Steel Water Bottle",
    "Organic Cotton Bedsheet Set",
    "Portable Power Bank 20000mAh",
    "Ergonomic Office Chair Deluxe"
]

units_sold = [120, 85, 60, 200, 45]

plt.figure(figsize=(6, 4))
bars = plt.bar(product_names, units_sold, color='skyblue')
plt.title('Units Sold per Product')
plt.xlabel('Product Name')
plt.ylabel('Number of Units Sold')
plt.xticks(rotation=90, ha='right')

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,height, f'{int(height)}', ha = 'center', va = 'bottom' )



plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.bar(product_names, units_sold, color='skyblue')
plt.title('Units Sold per Product')
plt.xlabel('Product Name')
plt.ylabel('Number of Units Sold')
plt.xticks(rotation=90, ha='right')
plt.tight_layout()
plt.savefig('test_chart_formatted1.png')
plt.show()
```
