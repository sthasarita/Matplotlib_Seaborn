```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import io

csv_data = """ProductID,ProductName,Category,Price,UnitsSold
101,Laptop,Electronics,1200,50
102,Mouse,Electronics,25,200
103,Keyboard,Electronics,75,150
104,T-Shirt,Apparel,20,300
105,Jeans,Apparel,95,120
106,Coffee Mug,Homeware,15,500
"""
# the csv_data is string type data and the format inside it is a csv format so importing io and using stringio will remove the string quotation of the csv_data
# and it makes that csv_data a filelike object from which the csv format data is being read and is being stored as  a dataframe

df = pd.read_csv(io.StringIO(csv_data)) # Load CSV data from a string using StringIO as a file-like object
print(df)
print(df["ProductName"])
```

```python
plt.figure(figsize = (10, 5))
plt.plot(df["ProductName"], df["Price"],label = "line1", marker = "1", linewidth= "1.8")
plt.title("ProductName vs Price")
plt.xlabel("ProductName")
plt.ylabel("Price")
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (10, 5))
plt.bar(df["ProductName"], df["Price"],label = "line1",color = "green")
plt.title("ProductName vs Price")
plt.xlabel("ProductName")
plt.ylabel("Price")
plt.legend()
plt.grid(True, alpha = 0.1)
plt.show()
```

```python
# Let's plot the number of units sold for each product 
plt.figure(figsize=(10, 6)) 

plt.bar(df['ProductName'], df['UnitsSold'], color='skyblue')

plt.title('Units Sold per Product')
plt.xlabel('Product Name')
plt.ylabel('Number of Units Sold')

plt.xticks(rotation=0, ha='right') # Rotate x-axis labels for readability

# "Ticks" are the little marks and text labels along an axis (in this case, the product
# names sitting along the x-axis).

plt.tight_layout() # Adjust plot to prevent labels from overlapping
plt.show()
```

```python

# introduces grid plt.suptitle
# plt.suptitle() = the heading at the very top of the page, like "Chapter 5: Data Visualization"
# plt.title() = the smaller heading above each individual exercise/question on that page

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Sample data - Monthly sales over a year
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales_2023 = [120, 135, 150, 165, 180, 200, 
              190, 210, 220, 230, 250, 300]
sales_2024 = [130, 145, 160, 170, 185, 210, 
              200, 220, 235, 245, 265, 320]

sales_2025 = [100, 1450, 180, 800, 100, 220, 
              200, 290, 235, 205, 400, 1320]

plt.figure(figsize=(12, 6))
plt.plot(months, sales_2023, marker='o', linewidth=2, label='2023 Sales', color='blue')
plt.plot(months, sales_2024, marker='s', linewidth=2, label='2024 Sales', color='red', linestyle=':')
plt.plot(months, sales_2025, marker=',', linewidth=2, label='2025 Sales', color='green', linestyle='--')

plt.suptitle('Monthly Sales Comparison (2023 vs 2024 vs 2025)', fontsize=14, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Sales (in thousands $)', fontsize=12)
plt.grid(True, alpha=1)
plt.legend()
plt.tight_layout()
plt.show()
```

```python

# tips

tips = sns.load_dataset("tips")
tips
```

```python
plt.scatter(tips["total_bill"], tips["tip"], s = tips["size"] * 20)
# so basically we can use scatter plot to represent the 3D data.
# As in this scatterplot we are utilizing this size(s) parameter of the scatter function to represent the size of the family had dinner in restro. 
# and obviously, there are total_bill and bill data also.
```

```python
# plt.plot vs plt.scatter
# plt.plot is useful for making scatterplot when we have large dataset containing crores of records 
# Because, the scatter function lags in such scenarios.But , we cann't use the features of scatter function such as size one (s)
# However, plt.scatter is useful to make scatter plot with all the functionality it provides for small dataset
```

```python
plt.plot(tips["total_bill"], tips["tip"], "o", markersize = tips["size"] * 20)

# Unlike plt.scatter(), plt.plot() only accepts a single scalar number (like markersize=10) because it renders all points as one single line object.
#     Passing a sequence of different sizes triggers TypeError: float() argument must be a string or a real number, not 'Series'.
```

```python
plt.figure(figsize = (6.4, 4.8))
scatter = plt.scatter(tips["total_bill"], tips["tip"], marker = "o", s = 50, c = tips["tip"], cmap = "viridis", alpha = 0.7, edgecolors = "black")
plt.title("Comparision of Total_bill and Tips")
plt.xlabel("total_bill")
plt.ylabel("tips")
plt.colorbar(scatter, label = "tip")
plt.grid()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("batsman_season_record.csv")
```

```python
df
```

```python
plt.figure(figsize = (6.4, 4.8))
plt.bar(np.arange(df.shape[0])-0.21, df["2015"], width = 0.21)
plt.bar(np.arange(df.shape[0]), df["2016"], width = 0.21)
plt.bar(np.arange(df.shape[0])+0.21, df["2017"], width = 0.21)

plt.xticks(np.arange(df.shape[0]), df["batsman"], rotation = 360)

plt.show()
```

```python
# stacked bar chart
plt.figure(figsize = (6.4, 4.8))
plt.bar(df["batsman"], df["2015"], label = "2015")
plt.bar(df["batsman"], df["2016"], bottom = df["2015"], label = "2016")
plt.bar(df["batsman"], df["2017"], bottom = df["2015"] + df["2016"], label = "2017")
plt.legend()
plt.show()
```
