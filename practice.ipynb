```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import io

csv_data = """ProductID,ProductName,Category,Price,UnitsSold
101,Laptop,Electronics,1200,50
102,Mouse,Electronics,25,200
103,Keyboard,Electronics,75,150
104,T-Shirt,Apparel,20,300
105,Jeans,Apparel,95,120
106,Coffee Mug,Homeware,15,500
"""
# the csv_data is string type data and the format inside it is a csv format so importing io and using stringio will remove the string quotation of the csv_data
# and it makes that csv_data a filelike object from which the csv format data is being read and is being stored as  a dataframe

df = pd.read_csv(io.StringIO(csv_data)) # Load CSV data from a string using StringIO as a file-like object
print(df)
print(df["ProductName"])
```

```python
plt.figure(figsize = (10, 5))
plt.plot(df["ProductName"], df["Price"],label = "line1", marker = "1", linewidth= "1.8")
plt.title("ProductName vs Price")
plt.xlabel("ProductName")
plt.ylabel("Price")
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (10, 5))
plt.bar(df["ProductName"], df["Price"],label = "line1",color = "green")
plt.title("ProductName vs Price")
plt.xlabel("ProductName")
plt.ylabel("Price")
plt.legend()
plt.grid(True, alpha = 0.1)
plt.show()
```

```python
# Let's plot the number of units sold for each product 
plt.figure(figsize=(10, 6)) 

plt.bar(df['ProductName'], df['UnitsSold'], color='skyblue')

plt.title('Units Sold per Product')
plt.xlabel('Product Name')
plt.ylabel('Number of Units Sold')

plt.xticks(rotation=0, ha='right') # Rotate x-axis labels for readability

# "Ticks" are the little marks and text labels along an axis (in this case, the product
# names sitting along the x-axis).

plt.tight_layout() # Adjust plot to prevent labels from overlapping
plt.show()
```

```python
# plt.plot vs plt.scatter
# plt.plot is useful for making scatterplot when we have large dataset containing crores of records 
# Because, the scatter function lags in such scenarios.But , we cann't use the features of scatter function such as size one (s)
# However, plt.scatter is useful to make scatter plot with all the functionality it provides for small dataset
```

```python
plt.plot(tips["total_bill"], tips["tip"], "o", markersize = tips["size"] * 20)

# Unlike plt.scatter(), plt.plot() only accepts a single scalar number (like markersize=10) because it renders all points as one single line object.
#     Passing a sequence of different sizes triggers TypeError: float() argument must be a string or a real number, not 'Series'.
```
