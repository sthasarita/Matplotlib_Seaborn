```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = ["sun", "mon", "tues", "wed"]
y = [24.6,25.6,30.6,19]
```

```python
plt.figure(figsize = (5, 5))
plt.plot(x, y, linestyle = "dashed", color = "indigo", linewidth = 3, marker = "*", markersize = 13)
plt.xlabel("days")
plt.ylabel("temperature")
plt.title("Analysis of temperature over days")
plt.grid()
plt.show()
```

```python
df  = pd.read_csv("batsman_season_record.csv")
```

```python
df.info()
```

```python
plt.barh(df["batsman"], df["2015"], color = "pink")

plt.show()
```

```python
plt.bar(np.arange(df.shape[0]) - 0.3, df["2015"], color = "pink", width = 0.3)
plt.bar(np.arange(df.shape[0]) , df["2016"], width = 0.3)
plt.bar(np.arange(df.shape[0]) + 0.3, df["2017"], width = 0.3)
plt.xticks(np.arange(df.shape[0]), df["batsman"], rotation = "vertical")

plt.show()
```

```python
plt.bar(df["batsman"], df["2015"], label = "2015")
plt.bar(df["batsman"], df["2016"], bottom = df["2015"], label = "2016")
plt.bar(df["batsman"], df["2017"], bottom = df["2016"]+ df["2015"], label = "2017")
plt.legend()
plt.show()
```
